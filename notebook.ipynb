## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

## Exploratory Data Analysis (EDA)

### Explore ...

```python
# Memeriksa struktur dataset untuk analisis penjualan
print("===== ORDERS =====")
display(orders_df.head())

print("===== ORDER ITEMS =====")
display(order_items_df.head())

print("===== PRODUCTS =====")
display(products_df.head())

print("===== PRODUCT CATEGORY TRANSLATION =====")
display(product_category_name_translation_df.head())
```

```python
# Menggabungkan data pesanan dengan item produk
sales_df = order_items_df.merge(
    orders_df[['order_id', 'order_purchase_timestamp', 'order_status']],
    on='order_id',
    how='inner'
)

# Menggabungkan dengan informasi kategori produk
sales_df = sales_df.merge(
    products_df[['product_id', 'product_category_name']],
    on='product_id',
    how='left'
)

# Menggabungkan dengan terjemahan kategori produk
sales_df = sales_df.merge(
    product_category_name_translation_df,
    on='product_category_name',
    how='left'
)

# Mengubah kategori yang belum memiliki terjemahan
sales_df['product_category_name_english'] = (
    sales_df['product_category_name_english']
    .fillna('unknown')
)

print("Jumlah baris data penjualan:", len(sales_df))
print("\nContoh hasil penggabungan:")
display(sales_df.head())
```

```python
# Membuat kolom periode bulanan
sales_df['order_month'] = (
    sales_df['order_purchase_timestamp'].dt.to_period('M')
)

# Menghitung total penjualan setiap bulan berdasarkan kategori produk
monthly_sales = (
    sales_df.groupby(
        ['order_month', 'product_category_name_english']
    )['price']
    .sum()
    .reset_index()
)

# Mengubah periode menjadi format tanggal
monthly_sales['order_month'] = (
    monthly_sales['order_month'].dt.to_timestamp()
)

print("Jumlah data penjualan bulanan:", len(monthly_sales))

print("\nContoh hasil perhitungan:")
display(monthly_sales.head(10))
```

```python
# Menghitung total penjualan setiap kategori produk
category_sales = (
    sales_df.groupby('product_category_name_english')['price']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

# Menghitung persentase kontribusi setiap kategori
total_sales = category_sales['price'].sum()

category_sales['sales_percentage'] = (
    category_sales['price'] / total_sales * 100
)

print("10 kategori produk dengan penjualan tertinggi:")
display(category_sales.head(10))

print("\nTotal seluruh nilai penjualan:",
      round(total_sales, 2))
```

```python
# Mengambil data penjualan dari pesanan yang berhasil dikirim
sales_delivered_df = sales_df[
    sales_df['order_status'] == 'delivered'
].copy()

# Menghitung total penjualan setiap kategori produk
category_sales = (
    sales_delivered_df
    .groupby('product_category_name_english')['price']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

# Menghitung persentase kontribusi setiap kategori
total_sales = category_sales['price'].sum()

category_sales['sales_percentage'] = (
    category_sales['price'] / total_sales * 100
)

print("Jumlah item dari pesanan delivered:", len(sales_delivered_df))
print("\n10 kategori produk dengan penjualan tertinggi:")
display(category_sales.head(10))

print("\nTotal nilai penjualan pesanan delivered:",
      round(total_sales, 2))
```

```python
# Mengelompokkan nilai penjualan berdasarkan bulan dan kategori produk
monthly_sales = (
    sales_delivered_df
    .groupby([
        sales_delivered_df['order_purchase_timestamp'].dt.to_period('M'),
        'product_category_name_english'
    ])['price']
    .sum()
    .reset_index()
)

# Mengubah periode bulan menjadi format tanggal
monthly_sales['order_purchase_timestamp'] = (
    monthly_sales['order_purchase_timestamp'].dt.to_timestamp()
)

# Mengganti nama kolom agar lebih mudah dipahami
monthly_sales.rename(
    columns={'order_purchase_timestamp': 'order_month'},
    inplace=True
)

print("Jumlah data penjualan bulanan berdasarkan kategori:",
      len(monthly_sales))

print("\nContoh hasil agregasi:")
display(monthly_sales.head(15))
```

```python
# Menentukan status keterlambatan setiap pesanan
delivery_df['is_late'] = (
    delivery_df['order_delivered_customer_date'] >
    delivery_df['order_estimated_delivery_date']
)

# Mengelompokkan pesanan berdasarkan bulan pembelian
delivery_df['order_month'] = (
    delivery_df['order_purchase_timestamp'].dt.to_period('M')
)

# Menghitung total dan jumlah pesanan terlambat setiap bulan
monthly_delivery = (
    delivery_df.groupby('order_month')
    .agg(
        total_orders=('order_id', 'count'),
        late_orders=('is_late', 'sum')
    )
    .reset_index()
)

# Menghitung persentase keterlambatan
monthly_delivery['late_percentage'] = (
    monthly_delivery['late_orders'] /
    monthly_delivery['total_orders'] * 100
)

# Mengubah periode menjadi format tanggal
monthly_delivery['order_month'] = (
    monthly_delivery['order_month'].dt.to_timestamp()
)

print("Jumlah periode pengiriman yang dianalisis:",
      len(monthly_delivery))

print("\nHasil persentase keterlambatan setiap bulan:")
display(monthly_delivery.head(15))

print("\nRata-rata persentase keterlambatan:",
      round(monthly_delivery['late_percentage'].mean(), 2), "%")
```

## Analisis Lanjutan (Opsional)

### Analisis RFM (Recency, Frequency, Monetary)

Analisis RFM digunakan untuk mengelompokkan pelanggan berdasarkan perilaku pembelian mereka melalui tiga indikator:

1. **Recency:** Mengukur jarak waktu sejak transaksi terakhir pelanggan.
2. **Frequency:** Mengukur jumlah transaksi yang dilakukan pelanggan.
3. **Monetary:** Mengukur total nilai pembelian pelanggan.

Analisis ini bertujuan untuk mengidentifikasi karakteristik pelanggan berdasarkan aktivitas dan kontribusi pembeliannya. Hasilnya dapat digunakan sebagai dasar penyusunan strategi pemasaran dan peningkatan loyalitas pelanggan.

```python
# Menggabungkan data transaksi

# Mengambil pesanan yang berstatus delivered
delivered_orders = orders_df[
    orders_df['order_status'] == 'delivered'
][[
    'order_id',
    'customer_id',
    'order_purchase_timestamp'
]].copy()

# Menghitung total nilai pembelian setiap pesanan
order_revenue = (
    order_items_df.groupby('order_id')['price']
    .sum()
    .reset_index()
    .rename(columns={'price': 'monetary'})
)

# Menghubungkan pesanan dengan identitas pelanggan unik
rfm_data = delivered_orders.merge(
    customers_df[['customer_id', 'customer_unique_id']],
    on='customer_id',
    how='inner'
)

# Menggabungkan nilai pembelian ke data pelanggan
rfm_data = rfm_data.merge(
    order_revenue,
    on='order_id',
    how='inner'
)

# Memeriksa hasil penggabungan
print("Jumlah transaksi:", len(rfm_data))
print("Jumlah pelanggan unik:", rfm_data['customer_unique_id'].nunique())

display(rfm_data.head())
```

```python
# Menghitung Recency, Frequency, dan Monetary

# Menentukan tanggal acuan analisis
snapshot_date = (
    rfm_data['order_purchase_timestamp'].max()
    + pd.Timedelta(days=1)
)

# Menghitung RFM untuk setiap pelanggan unik
rfm_df = (
    rfm_data.groupby('customer_unique_id')
    .agg(
        recency=(
            'order_purchase_timestamp',
            lambda x: (snapshot_date - x.max()).days
        ),
        frequency=('order_id', 'nunique'),
        monetary=('monetary', 'sum')
    )
    .reset_index()
)

# Memeriksa hasil perhitungan RFM
print("Jumlah pelanggan dalam analisis RFM:", len(rfm_df))
print("\nStatistik RFM:")
display(rfm_df[['recency', 'frequency', 'monetary']].describe())

print("\nContoh hasil RFM:")
display(rfm_df.head(10))
```

```python
# Membuat skor RFM

# Membuat skor Recency (nilai lebih kecil = skor lebih tinggi)
rfm_df['R_score'] = pd.qcut(
    rfm_df['recency'].rank(method='first'),
    4,
    labels=[4, 3, 2, 1]
).astype(int)

# Membuat skor Frequency (nilai lebih besar = skor lebih tinggi)
rfm_df['F_score'] = pd.qcut(
    rfm_df['frequency'].rank(method='first'),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

# Membuat skor Monetary (nilai lebih besar = skor lebih tinggi)
rfm_df['M_score'] = pd.qcut(
    rfm_df['monetary'].rank(method='first'),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

# Menggabungkan ketiga skor RFM
rfm_df['RFM_score'] = (
    rfm_df['R_score'].astype(str) +
    rfm_df['F_score'].astype(str) +
    rfm_df['M_score'].astype(str)
)

print("Contoh hasil pemberian skor RFM:")
display(rfm_df.head(10))

print("\nDistribusi skor RFM:")
display(rfm_df[['R_score', 'F_score', 'M_score']].describe())
```

```python
# Memperbaiki skor Frequency

# Memperbaiki skor Frequency berdasarkan jumlah transaksi
rfm_df['F_score'] = pd.cut(
    rfm_df['frequency'],
    bins=[0, 1, 2, 3, float('inf')],
    labels=[1, 2, 3, 4]
).astype(int)

# Memperbarui gabungan skor RFM
rfm_df['RFM_score'] = (
    rfm_df['R_score'].astype(str) +
    rfm_df['F_score'].astype(str) +
    rfm_df['M_score'].astype(str)
)

# Memeriksa distribusi skor Frequency
print("Distribusi skor Frequency:")
display(rfm_df['F_score'].value_counts().sort_index())

print("\nContoh hasil skor RFM setelah diperbaiki:")
display(rfm_df.head(10))
```

```python
# Membuat segmentasi pelanggan

# Membuat fungsi segmentasi pelanggan
def segment_customer(row):
    if row['R_score'] >= 3 and row['F_score'] >= 3 and row['M_score'] >= 3:
        return 'Champions'

    elif row['F_score'] >= 3 and row['R_score'] >= 2:
        return 'Loyal Customers'

    elif row['R_score'] >= 3 and row['F_score'] <= 2:
        return 'Potential Loyalists'

    elif row['R_score'] <= 2:
        return 'At Risk'

    else:
        return 'Others'


# Menerapkan segmentasi pada setiap pelanggan
rfm_df['customer_segment'] = rfm_df.apply(
    segment_customer,
    axis=1
)

# Menghitung jumlah pelanggan pada setiap segmen
segment_summary = (
    rfm_df['customer_segment']
    .value_counts()
    .reset_index()
)

segment_summary.columns = ['customer_segment', 'total_customers']

# Menghitung persentase setiap segmen
segment_summary['percentage'] = (
    segment_summary['total_customers'] /
    segment_summary['total_customers'].sum() * 100
)

print("Hasil segmentasi pelanggan:")
display(segment_summary)
```

```python
# Visualisasi segmentasi pelanggan

# Mengurutkan segmen pelanggan berdasarkan jumlah pelanggan
segment_order = [
    'At Risk',
    'Potential Loyalists',
    'Loyal Customers',
    'Champions'
]

segment_visual = (
    segment_summary
    .set_index('customer_segment')
    .reindex(segment_order)
    .reset_index()
)

# Membuat diagram batang horizontal
plt.figure(figsize=(12, 6))

bars = plt.barh(
    segment_visual['customer_segment'],
    segment_visual['total_customers']
)

# Menambahkan label jumlah dan persentase
for bar, total, percentage in zip(
    bars,
    segment_visual['total_customers'],
    segment_visual['percentage']
):
    plt.text(
        bar.get_width() + 500,
        bar.get_y() + bar.get_height() / 2,
        f'{total:,} ({percentage:.2f}%)',
        va='center',
        fontsize=10
    )

plt.title(
    'Distribusi Pelanggan Berdasarkan Segmen RFM',
    fontsize=14,
    fontweight='bold'
)
plt.xlabel('Jumlah Pelanggan')
plt.ylabel('Segmen Pelanggan')
plt.grid(axis='x', linestyle='--', alpha=0.4)
plt.xlim(0, segment_visual['total_customers'].max() * 1.25)
plt.tight_layout()
plt.show()
```
